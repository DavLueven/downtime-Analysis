```python
def replace_whitespace(string, length):
    result = ""
    for c in string[:length]:
        if c != " ":
            result +=  c
        else:
            result += "%20"
    return result
```

```python
string = "code academy.com   "
length = len(string.strip())

replace_whitespace(string, length)
```

```python
def urify(string, length):
    parts = []
    for c in string[:length]:
        if c != " ":
            parts.append(c)
        else:
            parts.append("%20")
    return "".join(parts)
```

```python
string1 = "code academy.com   "
length1 = len(string1.rstrip())

urify(string1, length1)
```

```python
string2 = "option strat.com   "
length2 = len(string2.rstrip())

urify(string2, length2)
```
